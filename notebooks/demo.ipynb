## Conclusion

Overall, the fine-tuned DeBERTa model showed exceptional performance during evaluation. While these scores confirm that the model successfully generalizes to unseen entities thanks to a leakage-free split, an F1 score of 0.98 on a finite sample (169 entities) should be interpreted as "no errors observed in this specific test set" rather than absolute perfection. 

To further increase confidence in the model, the next step should be to make the dataset bigger. Just like in statistics, where a larger sample size gives more reliable results (a lower p-value), more data will help prove the model's stability. A bigger dataset will cover a wider variety of sentence structures and rare cases.

It is important to emphasize that this implementation is not positioned as an ultimate, universal solution for production environments, but rather as a robust Proof-of-Concept (PoC) demonstrating that this architectural and pipeline approach is viable and has a rightful place in solving domain-specific NER tasks.

A deeper look into the error analysis reveals specific structural challenges that the model faces:

-- Most of our naming errors happen with mountains containing words like del (such as Ojos del Salado). These foreign prepositions or articles are likely very rare or missing in the training set, which is why the model struggles to capture the full name.

-- When a mountain name appears in a sentence with a strong metaphorical context (like "a mountain of unread emails"), the model gets confused by the surrounding text and completely misses the real entity.
